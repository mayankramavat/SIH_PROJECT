# National LMIS — 01: Raw Data Exploration & Profiling
**Ministry of Skill Development & Entrepreneurship (MSDE) | SIH_26246**

This notebook explores and profiles all raw labor market data files in `data/raw/`:
1. `job_postings.csv`: Historical demand postings by state, district, sector, and trade.
2. `training_data.csv`: Institutional capacity, enrollment, and certified trainee cohorts.
3. `placements.csv`: Post-training candidate placement records.
4. `districts.csv`: Master list of Indian states and districts.
5. `trades.csv`: Vocational trades mapped to NSQF levels and NCO-2015 codes.


```python
import pandas as pd
import numpy as np
from pathlib import Path

ROOT = Path('..').resolve()
RAW_DIR = ROOT / 'data' / 'raw'
print('Raw data directory:', RAW_DIR)
```

### 1. Ingest & Inspect Job Postings (Time Series Demand)


```python
df_jobs = pd.read_csv(RAW_DIR / 'job_postings.csv')
print(f'Shape: {df_jobs.shape}')
display(df_jobs.head())
print(df_jobs.info())
```

### 2. Ingest & Inspect Training Capacity Data (Supply)


```python
train_file = RAW_DIR / 'training_data.csv'
if not train_file.exists():
    train_file = RAW_DIR / 'training_capacity.csv'
df_train = pd.read_csv(train_file)
print(f'Shape: {df_train.shape}')
display(df_train.head())
```

### 3. Ingest & Inspect Placement Outcomes


```python
df_place = pd.read_csv(RAW_DIR / 'placements.csv')
print(f'Shape: {df_place.shape}')
display(df_place.head())
```

### 4. Master Metadata (Districts and Trades)


```python
df_dist = pd.read_csv(RAW_DIR / 'districts.csv')
df_trades = pd.read_csv(RAW_DIR / 'trades.csv')
print(f'Districts count: {len(df_dist)}, Trades count: {len(df_trades)}')
display(df_trades.head())
```

### 5. Check Missing Values Across All Datasets


```python
for name, df in [('Job Postings', df_jobs), ('Training Data', df_train), ('Placements', df_place)]:
    nulls = df.isnull().sum()
    total_nulls = nulls.sum()
    print(f'{name:<15}: {total_nulls} missing values across {len(df)} records.')
```

### Key Exploration Insights:
- 480 monthly vacancy records spanning 17 districts and 12 vocational trades.
- Training and placement datasets are cleanly aligned by trade, district, and sector.
- No unhandled NaN values in raw keys.
